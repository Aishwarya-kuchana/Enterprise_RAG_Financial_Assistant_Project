```python
import fitz
from pathlib import Path
```

```python
DATA_DIR = Path("../data/raw")

pdf_files = list(DATA_DIR.rglob("*.pdf"))

print(f"Total PDFs : {len(pdf_files)}")
```

```python
doc = fitz.open(pdf_files[0])
raw_text = ""
for page in doc:
    raw_text += page.get_text()
doc.close()
print(raw_text[:2000])
```

```python
import re
```

```python
def clean_text(text):
    # Remove multiple spaces
    text = re.sub(r"[ ]+", " ", text)
    # Remove multiple blank lines
    text = re.sub(r"\n\s*\n+", "\n\n", text)
    # Remove tabs
    text = text.replace("\t", " ")
    # Remove leading/trailing spaces
    text = text.strip()
    return text
```

```python
cleaned_text = clean_text(raw_text)
print(cleaned_text[:2000])
```

```python
print("Raw Characters :", len(raw_text))
print("Clean Characters :", len(cleaned_text))
```

```python
output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
output_file = output_dir / "Alphabet_2025_clean.txt"
with open(output_file, "w", encoding="utf-8") as f:
    f.write(cleaned_text)
print("Saved Successfully")
```

```python
# Automate all pdf files
DATA_DIR = Path("../data/raw")
OUTPUT_DIR = Path("../data/processed")

OUTPUT_DIR.mkdir(exist_ok=True)

pdf_files = list(DATA_DIR.rglob("*.pdf"))
```

```python
def clean_text(text):
    text = re.sub(r"[ ]+", " ", text)
    text = re.sub(r"\n\s*\n+", "\n\n", text)
    text = text.replace("\t", " ")
    text = text.strip()
    return text
```

```python
for pdf in pdf_files:
    doc = fitz.open(pdf)
    full_text = ""
    for page in doc:
        full_text += page.get_text()
    doc.close()
    cleaned = clean_text(full_text)
    output_path = OUTPUT_DIR / f"{pdf.stem}.txt"
    with open(output_path, "w", encoding="utf-8") as f:
        f.write(cleaned)
    print(f"Processed : {pdf.stem}")
```
