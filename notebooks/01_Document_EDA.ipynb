```python
import pandas as pd
from pathlib import Path
import fitz
```

```python
DATA_DIR = Path("../data/raw")
pdf_files = list(DATA_DIR.rglob("*.pdf"))
print(f"Total PDFs found: {len(pdf_files)}")
for pdf in pdf_files:
    print(pdf)
```

```python
data = []
for pdf in pdf_files:
    data.append({
        "Company": pdf.parent.name,
        "File Name": pdf.name,
        "Full Path": str(pdf),
        "File Size (MB)": round(pdf.stat().st_size / (1024 * 1024), 2)
    })
df = pd.DataFrame(data)
df
```

```python
df.sort_values(
    by="File Size (MB)",
    ascending=False
)
```

```python
df = df[
    [
        "Company",
        "File Name",
        "File Size (MB)",
        "Full Path"
    ]
]

df
```

```python
document_info = []
for pdf in pdf_files:
    doc = fitz.open(pdf)
    document_info.append({
        "Company": pdf.parent.name,
        "Pages": len(doc),
        "PDF Format": doc.metadata.get("format"),
        "Encrypted": doc.metadata.get("encryption"),
        "Creator": doc.metadata.get("creator")
    })
    doc.close()
document_df = pd.DataFrame(document_info)
document_df
```

```python
# Open the first document
doc = fitz.open(pdf_files[0])

# Read the first page
page = doc.load_page(4)

# Extract text
text = page.get_text()

print(repr(text[:1000]))

print(doc)
doc.close()
```

```python
# counting number of characters each pag of a file has
page_stats = []
doc = fitz.open(pdf_files[0])
for page_num in range(len(doc)):
    page = doc.load_page(page_num)
    text = page.get_text()
    page_stats.append({
        "Page": page_num + 1,
        "Characters": len(text),
        "Words": len(text.split())
    })
doc.close()
page_df = pd.DataFrame(page_stats)
page_df.head(10)
```

```python
print("Total Pages :", len(page_df))
print("Total Characters :", page_df["Characters"].sum())
print("Total Words :", page_df["Words"].sum())
print("\nAverage Characters/Page :", round(page_df["Characters"].mean(),2))
print("Average Words/Page :", round(page_df["Words"].mean(),2))
print("\nMaximum Characters on a Page :", page_df["Characters"].max())
print("Minimum Characters on a Page :", page_df["Characters"].min())
```

```python
empty_pages = page_df[page_df["Words"] == 0]
empty_pages
```
