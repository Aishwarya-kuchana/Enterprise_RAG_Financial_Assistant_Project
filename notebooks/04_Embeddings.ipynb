```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
print("Embedding model loaded successfully!")
```

```python
# model Testing 
text = "Alphabet invested heavily in AI infrastructure."
embedding = embedding_model.embed_query(text)
print("Embedding Length:", len(embedding))
```

```python
text1 = "Alphabet invested in AI."
text2 = "Google spent money on artificial intelligence."
emb1 = embedding_model.embed_query(text1)
emb2 = embedding_model.embed_query(text2)
print(len(emb1))
print(len(emb2))
```
